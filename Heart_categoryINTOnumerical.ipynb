```python
import pandas as pd

```

```python
data = pd.read_csv("heart.csv")
```

```python
categorical_columns = [
    "sex",
    "cp",
    "fbs",
    "restecg",
    "exang",
    "slope",
    "ca",
    "thal"
]
```

```python
category_maps = {}
```

```python
def manual_encoding(data, columns, category_maps):

    encoded_data = data.copy()

    for column in columns:

        if column not in category_maps:
            category_maps[column] = {}

        encoded_values = []

        for value in encoded_data[column]:

           
            if value not in category_maps[column]:

                new_number = len(category_maps[column])

                category_maps[column][value] = new_number

            encoded_values.append(
                category_maps[column][value]
            )

        encoded_data[column] = encoded_values

    return encoded_data, category_maps
```

```python
data, category_maps = manual_encoding(
    data,
    categorical_columns,
    category_maps
)
```
