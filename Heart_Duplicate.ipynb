```python
import pandas as pd

```

```python
data = pd.read_csv("heart.csv")
```

```python
def compare_rows(index1, index2):

    for column in data.columns:

        if data.iloc[index1][column] != data.iloc[index2][column]:
            return False

    return True
```

```python
duplicate_indexes = []

for i in range(len(data)):

    for j in range(i):

        if compare_rows(i, j):

            duplicate_indexes.append(i)

            break
```

```python
print("Total duplicate rows:", len(duplicate_indexes))
```

```python
clean_data = data.drop(index=duplicate_indexes)
clean_data = clean_data.reset_index(drop=True)
```
