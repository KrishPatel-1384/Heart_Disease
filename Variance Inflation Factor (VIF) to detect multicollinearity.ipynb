```python
import pandas as pd
import numpy as np
```

```python


df = pd.read_csv("/home/patel-krish/anaconda3/BrainyBeam/heart.csv")
```

```python

X = df.drop(columns=["target"]).astype(float)

results = []

```

```python
for feature in X.columns:

    y = X[feature].to_numpy()

    other_features = X.drop(columns=[feature]).to_numpy()

    A = np.column_stack([
        np.ones(len(other_features)),
        other_features
    ])

  
```

```python
beta = np.linalg.pinv(A.T @ A) @ A.T @ y

y_pred = A @ beta

ss_res = np.sum((y - y_pred) ** 2)
ss_tot = np.sum((y - np.mean(y)) ** 2)

r_squared = 1 - (ss_res / ss_tot)


vif = 1 / (1 - r_squared)

results.append({
    "Feature": feature,
    "R_squared": r_squared,
    "VIF": vif
    })

```

```python
vif_results = pd.DataFrame(results)
print(vif_results.round(4).to_string(index=False))
```

```python
print("\nInterpretation:")

for _, row in vif_results.iterrows():
    if row["VIF"] > 10:
        status = "Serious multicollinearity"
    elif row["VIF"] > 5:
        status = "Potential multicollinearity"
    else:
        status = "Acceptable"

    print(f'{row["Feature"]}: VIF = {row["VIF"]:.4f} - {status}')
```
