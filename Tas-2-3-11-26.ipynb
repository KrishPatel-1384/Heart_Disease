## Case Study: Explainability of a Heart Disease Prediction Model

### 1. Introduction

This case study evaluates the interpretability of a machine learning model developed to predict the possibility of heart disease. The objective is not only to measure predictive performance but also to understand how the model reaches its predictions.

Random Forest was selected as the primary model for this interpretability analysis. Permutation importance, SHAP-style feature contributions, and LIME-style local explanations were used to investigate the influence of individual features.

The analysis focuses on understanding both global model behaviour and individual predictions.

### 2. Qualitative Domain-Expert Perspectives

The following are illustrative perspectives that should be validated through discussions with qualified healthcare professionals.

**Perspective 1: Clinical relevance**

A healthcare professional would generally expect features such as chest pain characteristics, maximum heart rate, age, blood pressure, and cholesterol to be considered in the context of a patient's overall clinical condition.

**Perspective 2: Individual patient differences**

Two patients with similar medical measurements may receive different predictions because of differences in their other clinical characteristics. Local explanations can help identify which features contributed to each prediction.

**Perspective 3: Trust and transparency**

Healthcare professionals need understandable explanations rather than prediction scores alone. They should be able to identify the important factors behind a prediction and understand the limitations of the model.

**Perspective 4: Risk of misleading explanations**

An explanation does not establish that a feature causes heart disease. Feature importance describes model behaviour, not a medical diagnosis or causal relationship.

### 3. Findings from the Interpretability Analysis

The global feature-importance analysis identifies features that have a strong influence on the model across the evaluated dataset.

Permutation importance measures how model performance changes when a feature is shuffled. SHAP-style analysis estimates feature contributions to individual predictions and summarizes their magnitude across observations. LIME-style analysis approximates the model locally around a selected observation.

The three techniques provide different perspectives. Agreement between them can increase confidence that a feature is influential, while disagreement may indicate that its importance depends on the explanation method or individual patient.

### 4. Global and Local Interpretability

Global interpretability describes the overall behaviour of the model across the dataset. It helps identify features that frequently influence predictions.

Local interpretability explains an individual prediction. It helps investigate why the model assigned a particular output to a particular patient.

Both approaches are important. A feature may have high global importance but limited influence on a specific patient's prediction.

### 5. Limitations Identified

- Feature importance does not establish causation.
- SHAP and LIME explanations depend on their assumptions and implementation.
- Correlated features may share or distort apparent importance.
- Synthetic observations used by local explanation methods may not represent realistic patients.
- Explanations can vary between patients and model versions.
- A technically understandable explanation may still be medically incomplete.
- The dataset may not represent every population or clinical setting.

### 6. Proposed Improvements

- Validate explanations with qualified healthcare professionals.
- Use medically meaningful feature descriptions.
- Include confidence or uncertainty information alongside predictions.
- Test explanations across different patient groups.
- Investigate correlated features and unrealistic feature combinations.
- Compare multiple explanation methods instead of relying on one technique.
- Monitor explanation changes after model retraining.
- Maintain documentation of model versions, datasets, and evaluation results.

### 7. Conclusion

The case study demonstrates that model interpretability is an important part of developing responsible machine learning systems for healthcare-related applications.

Global methods help identify influential features across the dataset, while local methods help explain individual predictions. Combining these methods provides a more complete understanding of model behaviour.

However, explanations should be reviewed by qualified healthcare professionals before any real-world clinical use. The model should be treated as an educational prediction system rather than a replacement for medical judgement.
