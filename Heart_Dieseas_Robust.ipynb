```python
import numpy as np
import pandas as pd
```

```python
data = pd.read_csv("heart.csv")
```

```python
data
```

```python
 median_bps=data["trestbps"].median()
median_bps
```

```python
q1 = data["trestbps"].quantile(0.25)
q1
```

```python
q2 = data["trestbps"].quantile(0.75)
q2
```

```python
q = q2-q1
q
```

```python
data["trestbps_robust"] = (data["trestbps"] - median_bps) /q
q
```

```python
data["oldpeak_log"] = np.log1p(data["oldpeak"])
```

```python
a_chol = data['chol'].quantile(0.05)
a_chol
```

```python
b_chol = data['chol'].quantile(0.95)
b_chol
```

```python
data["chol_clipped"] = data["chol"].clip(lower=a_chol, upper=b_chol)
```

```python
min_chol = data["chol_clipped"].min()
max_chol = data["chol_clipped"].max()
```

```python
data["chol_winsor_minmax"] = (data["chol_clipped"] - min_chol) / (max_chol - min_chol)
```
