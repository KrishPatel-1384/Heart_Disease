```python
import pandas as pd
import numpy as np
import csv
```

```python
df = pd.read_csv("/home/patel-krish/anaconda3/BrainyBeam/heart.csv")
```

```python
rng = np.random.default_rng(42)
train_indices = []
test_indices = []

for label in df["target"].unique():
    indices = np.where(df["target"].to_numpy() == label)[0]
    rng.shuffle(indices)

    split_point = int(round(len(indices) * 0.80))

    train_indices.extend(indices[:split_point])
    test_indices.extend(indices[split_point:])
```

```python
rng.shuffle(train_indices)
rng.shuffle(test_indices)
```

```python
train_df = df.iloc[train_indices]
test_df = df.iloc[test_indices]
```

```python
def save_csv(data, filename):
    with open(filename, "w", encoding="utf-8", newline="") as file:
        writer = csv.writer(file, lineterminator="\n")
        writer.writerow(data.columns)

        for row in data.itertuples(index=False, name=None):
            writer.writerow([
                "" if pd.isna(value) else value
                for value in row
            ])
```

```python
save_csv(train_df, "heart_train.csv")
save_csv(test_df, "heart_test.csv")
```

```python
print("Training rows:", len(train_df))
print("Testing rows:", len(test_df))

print("\nOriginal class proportions:")
print(df["target"].value_counts(normalize=True).sort_index())

print("\nTraining class proportions:")
print(train_df["target"].value_counts(normalize=True).sort_index())

print("\nTesting class proportions:")
print(test_df["target"].value_counts(normalize=True).sort_index())

print("\nFiles saved successfully.")
```
