```python
import pandas as pd
```

```python
data = pd.read_csv("heart.csv")
```

```python
categorical_columns = [
    'sex',
    'cp',
    'fbs',
    'restecg',
    'exang',
    'slope',
    'ca',
    'thal'
]
```

```python
category_memory = {}
```

```python
def manual_one_hot_encode(data, columns, category_memory):

    encoded_data = data.copy()

    for col in columns:

        if col not in category_memory:
            category_memory[col] = []

        current_categories = data[col].dropna().unique()

        for category in current_categories:

            if category not in category_memory[col]:
                category_memory[col].append(category)

        for category in category_memory[col]:

            new_column = col + "_" + str(category)

            encoded_data[new_column] = (
                data[col] == category
            ).astype(int)

        encoded_data.drop(columns=[col], inplace=True)

    return encoded_data, category_memory
```

```python
encoded_df, category_memory = manual_one_hot_encode(
    data,
    categorical_columns,
    category_memory
)

encoded_df.head()
```
