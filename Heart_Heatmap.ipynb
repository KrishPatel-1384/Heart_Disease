```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
```

```python
data = pd.read_csv("/home/patel-krish/BrainyBeam/heart.csv")
```

```python
def check_missing(value):

    if value is None:
        return True

    if value == "":
        return True

    if value != value:
        return True

    return False
```

```python
missing_grid = []

for i in range(len(data)):

    row = []

    for column in data.columns:

        if check_missing(data.loc[i, column]):

            row.append(1)

        else:

            row.append(0)

    missing_grid.append(row)
```

```python
fig, ax = plt.subplots(figsize=(12, 7))

for row in range(len(missing_grid)):

    for column in range(len(data.columns)):

        value = missing_grid[row][column]

        if value == 0:

            color = (
                0.85,
                0.95,
                0.85
            )

        else:

            color = (
                1.0,
                0.25,
                0.25
            )


        box = Rectangle(
            (column, row),
            1,
            1,
            facecolor=color
        )

        ax.add_patch(box)


ax.set_xlim(
    0,
    len(data.columns)
)

ax.set_ylim(
    len(data),
    0
)


ax.set_xticks(
    range(len(data.columns))
)

ax.set_xticklabels(
    data.columns,
    rotation=90
)


ax.set_xlabel("Columns")

ax.set_ylabel("Rows")

ax.set_title(
    "Missing Values Heatmap"
)

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(12, 7))

for row in range(len(missing_grid)):

    for column in range(len(data.columns)):

        value = missing_grid[row][column]

        red = 0.2 + (0.8 * value)

        green = 0.9 - (0.7 * value)

        blue = 0.3

        color = (
            red,
            green,
            blue
        )


        box = Rectangle(
            (column, row),
            1,
            1,
            facecolor=color
        )

        ax.add_patch(box)


ax.set_xlim(
    0,
    len(data.columns)
)

ax.set_ylim(
    len(data),
    0
)

ax.set_xticks(
    range(len(data.columns))
)

ax.set_xticklabels(
    data.columns,
    rotation=90
)

ax.set_xlabel("Columns")

ax.set_ylabel("Rows")

ax.set_title(
    "Manually Created Missing Value Heatmap"
)

plt.show()
```
