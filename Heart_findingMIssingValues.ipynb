```python
import pandas as pd

```

```python
data = pd.read_csv("heart.csv")
```

```python
def check_missing(value):

    if value is None:
        return True

    if value == "":
        return True

    if value != value:
        return True

    return False
```

```python
missing_count = {}

for column in data.columns:

    count = 0

    for value in data[column]:

        if check_missing(value):
            count = count + 1

    missing_count[column] = count
```

```python
for column in missing_count:

    print(
        column,
        ":",
        missing_count[column]
    )
```
