```python
import pandas as pd
import matplotlib.pyplot as plt

```

```python
data = pd.read_csv("heart.csv")
```

```python
numerical_columns = data.select_dtypes(include=["int64", "float64"]).columns
numerical_columns
```

```python
for column in numerical_columns:

    plt.figure(figsize=(7, 3))

    plt.boxplot(
        data[column],
        vert=False
    )

    plt.title("Boxplot of " + column)
    plt.xlabel(column)

    plt.grid(alpha=0.3)

    plt.show()
```

```python
no_disease = data[data["target"] == 0]["chol"]

disease = data[data["target"] == 1]["chol"]

plt.figure(figsize=(8, 5))

plt.hist(
    no_disease,
    bins=15,
    alpha=0.5,
    edgecolor="black",
    label="No Heart Disease"
)

plt.hist(
    disease,
    bins=15,
    alpha=0.5,
    edgecolor="black",
    label="Heart Disease"
)

plt.title("Cholesterol Comparison")

plt.xlabel("Cholesterol")
plt.ylabel("Frequency")

plt.legend()
plt.grid(alpha=0.3)

plt.show()
```

```python
no_disease = data[data["target"] == 0]["thalach"]

disease = data[data["target"] == 1]["thalach"]

plt.figure(figsize=(8, 5))

plt.hist(
    no_disease,
    bins=15,
    alpha=0.5,
    edgecolor="black",
    label="No Heart Disease"
)

plt.hist(
    disease,
    bins=15,
    alpha=0.5,
    edgecolor="black",
    label="Heart Disease"
)

plt.title("Maximum Heart Rate Comparison")

plt.xlabel("Maximum Heart Rate")
plt.ylabel("Frequency")

plt.legend()
plt.grid(alpha=0.3)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    data["chol"],
    data["thalach"],
    alpha=0.5
)

plt.title("Cholesterol vs Maximum Heart Rate")

plt.xlabel("Cholesterol")
plt.ylabel("Maximum Heart Rate")

plt.grid(alpha=0.3)

plt.show()
```
