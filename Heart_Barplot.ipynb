```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("/home/patel-krish/BrainyBeam/heart.csv")
```

```python
categorical_columns = [
    "sex",
    "cp",
    "fbs",
    "restecg",
    "exang",
    "slope",
    "ca",
    "thal"
]
```

```python
def count_categories(values):

    counts = {}

    for value in values:

        if value in counts:
            counts[value] = counts[value] + 1

        else:
            counts[value] = 1

    return counts
```

```python
colors = [
    "skyblue",
    "orange",
    "lightgreen",
    "pink",
    "gold",
    "violet"
]

for column in categorical_columns:

    counts = count_categories(data[column])

    categories = list(counts.keys())
    frequencies = list(counts.values())


    if len(categories) <= 2:
        bar_width = 0.4

    elif len(categories) <= 4:
        bar_width = 0.6

    else:
        bar_width = 0.8


    bar_colors = []

    for i in range(len(categories)):

        bar_colors.append(
            colors[i % len(colors)]
        )


    plt.figure(figsize=(7, 5))


    bars = plt.bar(
        categories,
        frequencies,
        width=bar_width,
        color=bar_colors,
        edgecolor="black"
    )


    plt.title(
        "Bar Plot of " + column
    )

    plt.xlabel(column)

    plt.ylabel("Frequency")


    for i in range(len(categories)):

        plt.text(
            categories[i],
            frequencies[i] + 5,
            str(frequencies[i]),
            ha="center"
        )


    plt.grid(
        axis="y",
        alpha=0.3
    )

    plt.show()
```
