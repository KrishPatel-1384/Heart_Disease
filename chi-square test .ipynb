```python
chi_square = 0

for observed_value, expected_value in [
    (86, 151.83),
    (226, 160.17),
    (413, 347.17),
    (300, 365.83)
]:
    contribution = (
        (observed_value - expected_value) ** 2
    ) / expected_value

    print("Contribution:", contribution)

    chi_square += contribution

print("Chi-square value:", chi_square)
```
