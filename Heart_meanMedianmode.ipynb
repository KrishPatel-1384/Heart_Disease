```python
def mean(values):

    total = 0

    for value in values:
        total = total + value

    mean = total / len(values)

    return mean
```

```python
def manual_median(values):

    values = list(values)

    values.sort()

    n = len(values)

    if n % 2 == 0:

        middle1 = values[n // 2 - 1]
        middle2 = values[n // 2]

        median = (middle1 + middle2) / 2

    else:

        median = values[n // 2]

    return median
```

```python
def mode(values):

    count = {}

    for value in values:

        if value in count:
            count[value] = count[value] + 1

        else:
            count[value] = 1

    highest_count = 0
    mode_value = None

    for value in count:

        if count[value] > highest_count:

            highest_count = count[value]
            mode_value = value

    return mode_value
```

```python
def min(values):

    minimum = values.iloc[0]

    for value in values:

        if value < minimum:
            minimum = value

    return minimum
```

```python
def max(values):

    maximum = values.iloc[0]

    for value in values:

        if value > maximum:
            maximum = value

    return maximum
```

```python
def std(values):

    mean = manual_mean(values)

    total = 0

    for value in values:

        difference = value - mean

        total = total + difference ** 2

    variance = total / (len(values) - 1)

    std = variance ** 0.5

    return std
```
