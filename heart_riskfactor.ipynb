```python
import pandas as pd
```

```python
data = pd.read_csv("heart.csv")
```

```python
data["chol_percentile"] = data["chol"].rank(pct=True)


data["thalach_percentile"] = data["thalach"].rank(pct=True)
```

```python
data["low_thalach"] = 1 - data["thalach_percentile"]
```

```python
data["interaction"] = (
    data["chol_percentile"] *
    data["low_thalach"]
)
```

```python
data["risk_factor"] = (
    0.4 * data["chol_percentile"] +
    0.4 * data["low_thalach"] +
    0.2 * data["interaction"]
)
```
